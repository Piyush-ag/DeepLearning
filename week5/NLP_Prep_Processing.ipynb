# [Section 3] **Review**: Exploring the Preprocessing Pipeline

In this section, we will use a sample sentence to explore what happens at each step of preprocessing from Tokenization to Punctuation removal.

## 1. Tokenization
The process of converting text contained in paragraphs or sentences into individual words (called tokens) is known as tokenization.
This is usually a very important step in text preprocessing before we can convert text into vectors full of numbers.



```python
#sample_text = "He is retired.  He lives in Barnstable.  He likes to work in his yard.  He doesn't smoke."
sample_text = "Tesla makes a lot of money. Its sales went up by half. It says it delivered more cars and did not increase spending."
sample_text
```

```python
nlp = spacy.load("en_core_web_sm")
tokens = nlp(sample_text)
token_list = []
for token in tokens:
    token_list.append(token.text)
token_list
```

---

## 2. Lemmatization
Lemmatization (and stemming) attempts to get root word (for eg rain) for different word inflections (raining, rained etc).
Lemma algos gives you real dictionary words, whereas stemming simply cuts off last parts of the word so its faster but less accurate. Stemming returns words which are not really dictionary words and hence you will not be able to find pretrained vectors for it in Glove, Word2Vec etc and this is a major disadvantage depending on application.


```python
lemma_list = []
for word in tokens:
    #token_list.append(token.text)
    lemma = word.lemma_.strip()
    lemma_list.append(lemma)
#token_list
lemma = " ".join(lemma_list)
lemma
```

---
## 3. Stop Word removals (careful not to remove the negation)
There are certain words above such as “it”, “is”, “that”, “this” etc. which don’t contribute much to the meaning of the underlying sentence and are actually quite common across all English documents; these words are known as stop words. There is generally a need to remove these “common” words before vectorizing tokens by a count vectorizer so that we can reduce the total dimensions of our vectors, and mitigate the so called “curse of dimensionality”.


```python
stops_with_negation = stopwords.words('english')
stops = set(stopwords.words('english')) - set(["mustn't", 'won', "shouldn't", 'no', "mightn't", 'aren', 'doesn', "wasn't",  'mustn', "wouldn't", "shan't", "couldn't"\
                                               "hadn't", 'don', 'haven', "isn't", "haven't", "needn't", "aren't", 'shouldn', "don't", 'shan', "won't", 'hadn', \
                                               "weren't", 'didn', 'wouldn', "doesn't", 'not', "hasn't", 'isn', 'couldn', "didn't", 'ain', "couldn't", "hadn't", \
                                               'weren', 'mightn', 'needn'])

tokens = nltk.word_tokenize(lemma)




```

**Example of Removing Negation**

```python
words_no_negation = ' '.join([token for token in tokens if str(token) not in stops_with_negation])
words_no_negation
```

**Example Retaining Negation**

```python
#keep the negation
words_with_negation = ' '.join([token for token in tokens if str(token) not in stops])
words_with_negation
```

---
## 4. Removing Punctuation (however, be aware that sometimes punctuations are meaningful)
Once we have tokenized the text and have converted the word contractions it really isn't useful anymore to have punctuation and special characters in our text.



```python
tokens = nlp(words_with_negation)

punct_list = []
for token in tokens:
    if token.is_punct is False:
        punct_list.append(token.lemma_)

punct = " ".join(punct_list)
punct
```
